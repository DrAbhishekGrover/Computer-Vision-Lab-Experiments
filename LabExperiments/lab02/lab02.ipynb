```python
import numpy as np
import imageio.v2 as iio
import matplotlib.pyplot as plt

img=iio.imread('/content/sample_data/grayscale_arrow.tiff')
img2=np.copy(img)
theta=np.deg2rad(30)
sx,sy=1.5,1.5
tx,ty=100,50

R=np.array([[np.cos(theta),-np.sin(theta),0],[np.sin(theta),np.cos(theta),0],
            [0,0,1]])
S=np.array([[sx,0,0],[0,sy,0],[0,0,1]])
T=np.array([[1,0,tx],[0,1,ty],[0,0,1]])

isize=np.shape(img)
nrow=isize[0]
ncol=isize[1]

y,x=np.indices((nrow,ncol))
#Homogenous coordinate
X=np.vstack((x.ravel(),y.ravel(),np.ones(x.size)))
Y=R@X

x_new=np.round(Y[0,:]).astype(int)
y_new=np.round(Y[1,:]).astype(int)
valid=((x_new>=0)&(x_new<ncol)&(y_new>=0)&(y_new<nrow))

output_img=np.ones_like(img)*255
output_img[y_new[valid],x_new[valid]]=img2.ravel()[valid]
fig=plt.figure(figsize=(10,7))
nrow=1
ncol=2
fig.add_subplot(nrow,ncol,1)
plt.imshow(img,cmap='gray', vmax=255)
plt.axis('off')
plt.title('input')

fig.add_subplot(nrow,ncol,2)
plt.imshow(output_img,cmap='gray', vmax=255)
plt.axis('off')
plt.title('output')

plt.show()
```

```python
img=iio.imread('/content/sample_data/grayscale_arrow.tiff')
img2=np.copy(img)
theta=np.deg2rad(30)
sx,sy=1.5,1.5
tx,ty=100,50

R=np.array([[np.cos(theta),-np.sin(theta),0],[np.sin(theta),np.cos(theta),0],
            [0,0,1]])
S=np.array([[sx,0,0],[0,sy,0],[0,0,1]])
T=np.array([[1,0,tx],[0,1,ty],[0,0,1]])

isize=np.shape(img)
nrow=isize[0]
ncol=isize[1]
#nframe=isize[2]

y,x=np.indices((nrow,ncol))

#Inverse mapping
#Homogenous coordinate for new image
Y=np.vstack((x.ravel(),y.ravel(),np.ones(x.size)))
#Get old coordinate
X=np.linalg.inv(R)@Y

x_old=np.round(X[0,:]).astype(int)
y_old=np.round(X[1,:]).astype(int)

valid=((x_old>=0)&(x_old<nrow)&(y_old>=0)&(y_old<ncol))

#Create output image
output_image=np.ones_like(img)*255
output_image[y.ravel()[valid],x.ravel()[valid]]=img2[y_old[valid],x_old[valid]]

fig=plt.figure(figsize=(10,7))
nrow=1
ncol=2
fig.add_subplot(nrow,ncol,1)
plt.imshow(img,cmap='gray', vmax=255)
plt.axis('off')
plt.title('input')

fig.add_subplot(nrow,ncol,2)
plt.imshow(output_image,cmap='gray', vmax=255)
plt.axis('off')
plt.title('output')

plt.show()
```
